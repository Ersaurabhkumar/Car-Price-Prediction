```python
# Note to DevOps Team:-
# Below are the requirement files to be installed on the server.
import pandas as pd
import numpy as np
from pydantic import BaseModel
from sklearn.neighbors import KNeighborsRegressor
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from fastapi import FastAPI
import uvicorn
import nest_asyncio
```

```python
# Common Utils Functions
def categorical_encoding(data):
  for col in data.columns:
    if data[col].dtype=="object":
      label_encoder=LabelEncoder()
      data[col]=label_encoder.fit_transform(data[col])
  return data


def predictive_column(data,target):
  X=data.drop(target,axis=1)
  y=data[target]
  return X,y


def train_and_test_split(x,y,test_size=0.2,random_state=42):
  x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=test_size,random_state=random_state)
  return x_train,x_test,y_train,y_test
```

```python
# Code for the deployment
nest_asyncio.apply()

class InputData(BaseModel):
    features : list

df=pd.read_excel(r"C:\Users\saurabh kumar\Desktop\Institute\DE\Car Showroom Data.xlsx")
df.drop("TXNID",axis=1,inplace=True)
df = categorical_encoding(df)
x,y = predictive_column(df,'Amount')
x_train,x_test,y_train,y_test = train_and_test_split(x,y)
model = KNeighborsRegressor()
model.fit(x_train,y_train)
```

```python
# Code on the server
app = FastAPI()

@app.post('/car_amount_predictor')
def car_amount_predictor(data :InputData):
    feature_array = np.array(data.features).reshape(1,-1)
    prediction = model.predict(feature_array)
    return {'prediction':prediction.tolist()}
```

```python
# Hosting on the web link

uvicorn.run(app,host = '127.0.0.1', port = 8000)
```
